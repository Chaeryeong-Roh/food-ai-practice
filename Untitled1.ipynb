```python
# Wine 데이터셋 가져오기
wine_quality = fetch_ucirepo(id=186)
X_wine = wine_quality.data.features
y_wine = wine_quality.data.targets
df_wine = pd.concat([X_wine, y_wine], axis=1)
# Rice 데이터셋 가져오기
rice = fetch_ucirepo(id=545)
X_rice = rice.data.features
y_rice = rice.data.targets
df_rice = pd.concat([X_rice, y_rice], axis=1)
print("Wine shape:", df_wine.shape)
print("Rice shape:", df_rice.shape)
```

```python
def explain_tabular_dataset(df, label_col, dataset_name):
    feature_cols = [c for c in df.columns if c != label_col]
    print(f"\n### {dataset_name} ###")
    print("- sample: 데이터프레임의 한 행(row)")
    print("- label:", label_col)
    print("- feature 개수:", len(feature_cols))
    print("- feature 목록:", feature_cols)
```

```python
data = []

# Wine Quality 데이터셋 정보 추출
wine_label = "quality"
wine_features = [c for c in df_wine.columns if c != wine_label]
data.append({
    "데이터셋": "Wine Quality",
    "레이블": wine_label,
    "피처 개수": len(wine_features),
    "피처 목록": ", ".join(wine_features)
})

# Rice (Cammeo and Osmancik) 데이터셋 정보 추출
rice_label = "Class"
rice_features = [c for c in df_rice.columns if c != rice_label]
data.append({
    "데이터셋": "Rice (Cammeo and Osmancik)",
    "레이블": rice_label,
    "피처 개수": len(rice_features),
    "피처 목록": ", ".join(rice_features)
})

# 데이터프레임으로 만들고 출력
df_summary = pd.DataFrame(data)
display(df_summary)
```

```python
df_wine_cls = df_wine.copy()
def quality_to_class(q):
    if q <= 4:
        return "low"
    elif q <= 6:
        return "medium"
    else:
        return "high"
df_wine_cls["quality_class"] = df_wine_cls["quality"].apply(quality_to_class)
display(df_wine_cls[["quality", "quality_class"]].head())
print(df_wine_cls["quality_class"].value_counts())
```
